```python
# Gráfica
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Distribución inicial y matriz de transición
mu = np.array([0.5, 0.3, 0.2])
P = np.array([
    [0.2, 0.5, 0.3],
    [0.4, 0.0, 0.6],
    [0.5, 0.5, 0.0]
])

estados = ["a", "b", "c"]
colores = {"a": "#3678A8", "b": "#D9A441", "c": "#409080"}

def dibujar_barra(ax, probabilidades, etiqueta):
    """Dibuja intervalos acumulados en el orden a, b, c."""
    inicio = 0.0

    for estado, prob in zip(estados, probabilidades):
        if prob > 0:
            ax.barh(
                0, prob, left=inicio, height=0.55,
                color=colores[estado],
                edgecolor="white", linewidth=1.5
            )
            ax.text(
                inicio + prob / 2, 0, estado,
                ha="center", va="center",
                fontsize=14, fontweight="bold", color="white"
            )
        inicio += prob

    ax.set_xlim(0, 1)
    ax.set_ylim(-0.5, 0.5)
    ax.set_yticks([])
    ax.set_ylabel(etiqueta, rotation=0, labelpad=35,
                  fontsize=14, fontweight="bold")

    # Regla con marcas cada 0.1
    marcas = np.arange(0, 1.01, 0.1)
    ax.set_xticks(marcas)
    ax.set_xticklabels([f"{x:.1f}" for x in marcas])
    ax.tick_params(axis="x", labelsize=10, length=7)

    # Líneas verticales de referencia
    ax.grid(axis="x", linestyle="--", alpha=0.4)
    ax.set_axisbelow(True)

    for lado in ["top", "left", "right"]:
        ax.spines[lado].set_visible(False)


# Generar figura con las cuatro barras
fig, axes = plt.subplots(4, 1, figsize=(12, 7))

dibujar_barra(axes[0], mu, r"$\psi$")
dibujar_barra(axes[1], P[0], r"$\phi(a,\cdot)$")
dibujar_barra(axes[2], P[1], r"$\phi(b,\cdot)$")
dibujar_barra(axes[3], P[2], r"$\phi(c,\cdot)$")

fig.suptitle(
    "Funciones de iniciación y actualización",
    fontsize=16, fontweight="bold"
)

axes[-1].set_xlabel("Número aleatorio uniforme U", fontsize=12)

leyenda = [Patch(facecolor=colores[e], label=f"Estado {e}") for e in estados]
fig.legend(handles=leyenda, loc="lower center",
           ncol=3, bbox_to_anchor=(0.5, -0.02))

plt.tight_layout(rect=[0, 0.05, 1, 0.95])
plt.show()
```

**(a)** Con $\mu^{(0)}=\left(\frac12,\frac12\right)$, un estudiante programa $X_0=\psi(U_0)$ y $X_1=\phi(X_0,U_0)$, con el mismo $U_0$. Calcule $\mathbb{P}(X_1=s_1\mid X_0=s_1)$ y $\mathbb{P}(X_1=s_1\mid X_0=s_2)$, y describa en palabras qué relación hay entre $X_0$ y $X_1$. ¿Cuál debería ser la respuesta correcta? ¿Cuál es el error?

**Solución:**

Con $\mu^{(0)}=(\tfrac12,\tfrac12)$: $\psi(u)=s_1$ si $u\in[0,0.5)$ y $s_2$ si $u\in[0.5,1]$. Para Gotemburgo, $\phi(s_1,u)=s_1$ si $u\in[0,0.75)$ y $s_2$ si $u\in[0.75,1]$; $\phi(s_2,u)=s_1$ si $u\in[0,0.25)$ y $s_2$ si $u\in[0.25,1]$.

- Si $X_0=\psi(U_0)=s_1$ entonces $U_0\in[0,0.5)$. Como $[0,0.5)\subset[0,0.75)$, necesariamente $X_1=\phi(s_1, U_0)=s_1$, por lo cual
$$\mathbb{P}(X_1=s_1\mid X_0=s_1) = \frac{\mathbb{P}(U_0\in[0,0.5)\cap[0,0.75))}{\mathbb{P}(U_0\in[0,0.5))}=\frac{0.5}{0.5}=1.$$

- Si $X_0=\psi(U_0)=s_2$ entonces $U_0\in[0.5,1]$, que es disjunto de $[0,0.25)$, luego $X_1=\phi(s_2,U_0)=s_2$ y
$$\mathbb{P}(X_1=s_1\mid X_0=s_2) = \frac{\mathbb{P}(U_0\in[0.5,1]\cap[0,0.25))}{\mathbb{P}(U_0\in[0.5,1])}=0.$$

- **Relación entre $X_0$ y $X_1$:** al reutilizar $U_0$, $X_1$ queda completamente determinado por $X_0$ y de hecho $X_1=X_0$ con probabilidad $1$: el programa nunca cambia de estado en el primer paso.

- **Respuesta correcta:** con $U_1$ nuevo e independiente de $U_0$, $\mathbb{P}(X_1=s_1\mid X_0=s_1) = P_{11} = 0.75$ y $\mathbb{P}(X_1=s_1\mid X_0=s_2) = P_{21} = 0.25$.

- **El error:** usar el mismo $U_0$ para $\psi$ y para $\phi$. La construcción exige $X_1=\phi(X_0,U_1)$ con $U_1$ independiente de $U_0$ (y por tanto de $X_0$). Al reutilizarlo, saber $X_0$ ya restringe el valor de $U_0$ y el número usado en la transición deja de ser uniforme en $[0,1]$ dado $X_0$.

**(b)** Otro estudiante, empezando en lluvia, sortea un único $U$ y programa $X_{n+1}=\phi(X_n,U)$ para todo $n$. Describa todas las trayectorias posibles según el valor de $U$ y calcule $\mathbb{P}(X_2=s_1)$. Compárelo con el valor correcto $\frac12(1+2^{-2})$ del Problema 2.3. ¿Es $(X_n)$ una cadena de Markov con matriz $P$? ¿Cuál es el error?

**Solución:**

Tenemos $X_0=s_1$ y un único $U$ para todos los pasos. Veamos:

- Si $U\in[0,0.75)$: $X_1=\phi(s_1,U)=s_1$, y repitiendo, $X_n=s_1$ para todo $n$. Trayectoria: $s_1,s_1,s_1,\dots$
- Si $U\in[0.75,1]$: $X_1=\phi(s_1,U)=s_2$. Como $U\ge 0.75 \ge 0.25$, $X_2=\phi(s_2,U)=s_2$, y así $X_n=s_2$ para todo $n\ge1$. Trayectoria: $s_1,s_2,s_2,\dots$

Solo hay dos trayectorias posibles y ambas son constantes desde $n=1$. Por lo tanto
$$
\mathbb{P}(X_2=s_1)=\mathbb{P}(U\in[0,0.75))=0.75 .
$$

- El valor correcto es $\frac12(1+2^{-2})=\frac{5}{8}=0.625$. El programa da $0.75$, con una diferencia de $0.125$, así que la probabilidad obtenida es errónea.

- **¿Es $(X_n)$ una cadena de Markov con matriz $P$?** No. Por ejemplo, $\mathbb{P}(X_2=s_1\mid X_1=s_1)=1\neq P_{11}=0.75$ y $\mathbb{P}(X_2=s_2\mid X_1=s_2)=1\neq P_{22}=0.75$. El proceso sí resulta ser una cadena de Markov, pero **no homogénea**: la matriz de transición del paso $0\to1$ es $P$ y la de los pasos $n\to n+1$ con $n\ge1$ es la identidad $I$ (una vez dado el primer paso, el proceso se queda congelado). En particular, no tiene a $P$ como matriz de transición.

- **El error:** reutilizar el mismo número $U$ en todas las transiciones, en vez de generar un $U_{n+1}$ nuevo e independiente en cada paso. Así, el estado al que se llegó ya contiene información sobre $U$ y las transiciones posteriores dejan de ser independientes del pasado.

**(c)** Para el clima de Los Ángeles,

$$
P=\begin{pmatrix}
0.5 & 0.5\\
0.1 & 0.9
\end{pmatrix}
$$

un tercer estudiante construye la barra del estado $s_i$ con la columna $i$ de $P$ en lugar de la fila $i$, y su código devuelve el último estado cuando $x$ no cae en ningún intervalo. Determine las probabilidades de transición que realmente simula su programa. ¿Por qué el error pasa desapercibido en Gotemburgo? ¿Qué condición de una matriz de transición verifican que las filas sirven y las columnas en general no?




**Solución:**

Las columnas de $P$ son $(0.5,\ 0.1)$ (columna 1) y $(0.5,\ 0.9)$ (columna 2). Al usar la columna $i$ en vez de la fila $i$:

- Para el estado $s_1$ (columna 1, suma $0.6$): los intervalos son $s_1:[0,0.5)$ y $s_2:[0.5,0.6)$, y no cubren $[0.6,1]$. Como el código devuelve el último estado ($s_2$) cuando $x$ no cae en ningún intervalo, en la práctica:
  $$
  \phi(s_1,u)=
  \begin{cases}
  s_1,&u\in[0,\ 0.5)\\
  s_2,&u\in[0.5,\ 1]
  \end{cases}
  $$

- Para el estado $s_2$ (columna 2, suma $1.4$): los intervalos son $s_1:[0,0.5)$ y $s_2:[0.5,1.4)$. Como $u\in[0,1]$, el trozo $[1,1.4)$ nunca se alcanza, y en la práctica:
  $$
  \phi(s_2,u)=
  \begin{cases}
  s_1,&u\in[0,\ 0.5)\\
  s_2,&u\in[0.5,\ 1]
  \end{cases}
  $$

Luego la matriz de transición que realmente simula el programa es
$$
\hat P=\begin{pmatrix}
0.5 & 0.5\\
0.5 & 0.5
\end{pmatrix}
\qquad\text{en lugar de}\qquad
P=\begin{pmatrix}
0.5 & 0.5\\
0.1 & 0.9
\end{pmatrix}.
$$
La fila de $s_1$ coincide por casualidad, pero desde $s_2$ (sol) el programa pasa a lluvia con probabilidad $0.5$ en vez de $0.1$.

- **¿Por qué pasa desapercibido en Gotemburgo?** Porque esa matriz es simétrica, $P=P^T$: la columna $i$ es igual a la fila $i$, así que el programa construye exactamente las barras correctas.

- **¿Qué condición verifican las filas y no las columnas?** Una matriz de transición es estocástica por filas: $P_{ij}\ge 0$ y $\sum_j P_{ij}=1$ para cada $i$. Por eso los intervalos de una fila tienen longitudes que suman exactamente $1$ y parten $[0,1]$. Las columnas no tienen por qué sumar $1$ (aquí suman $0.6$ y $1.4$), así que sus intervalos pueden dejar huecos o salirse de $[0,1]$, que es justo lo que ocurrió.

$$
P=\begin{pmatrix}

0.40 & 0.20 & 0.10 & 0.30\\

## Ejercicio 1.
Sea $U_0 \sim U(0,1)$. Construya explicitamente una función de inicialización $\psi:[0,1] \to S$ tal que $X_0 = \psi(U_0)$ tenga la distribución inicial indicada.

```python
import numpy

# Semilla fija para resultados reproducibles.
numpy.random.seed(2002)

# probabilidades iniciales
def psi(u):
  if 0 <= u < 0.5:
    return 1
  elif 0.5 <= u < 0.75:
    return 2
  elif 0.75 <= u < 0.90:
    return 3
  else:
    return 4

u_0 = numpy.random.uniform(0, 1)
x_0 = psi(u_0)

print(f"Valor de la uniforme. u_0 = {u_0}")
print(f"Valor del estado. x_0 = {x_0}")
```

Explícitamente, acumulando $\mu^{(0)}=(0.50,0.25,0.15,0.10)$:
$$
\psi(u)=
\begin{cases}
1, & u\in[0,\ 0.50)\\
2, & u\in[0.50,\ 0.75)\\
3, & u\in[0.75,\ 0.90)\\
4, & u\in[0.90,\ 1]
\end{cases}
$$
Como $U_0\sim U(0,1)$, $\mathbb{P}(X_0=i)$ es la longitud del intervalo $i$: $0.50,\ 0.25,\ 0.15,\ 0.10$, que es la distribución inicial pedida.

Método de aproximación <br>
**Paso 1.** Fije el estado incial en $X_0 = 1$, es decir, suponga que el asegurado comienza saludable. <br>

**Paso 2.** Genere seis varibales aleatorias uniformes independientes: $U_1, U_2, \dots, U_6$. Utilizando la función de actualización $\phi$ construida en el punto $2$, simule $X_i = \phi (X_{i-1}, U_i)$, para $i = 1, \dots, 6$. <br>

**Paso 3.** Repita el procedimiento anterior $M = 1000$ veces, generando trayectorias independientes, todas comenzado en el estado $1$. Denote por $X_6^{(r)}$ el estado final de la trayectoria número $r$, para $r = 1, \dots, M$.

**Paso 4.** Cuente el número de trayectorias que terminana en el estado $3$:

\begin{align*}
N_3= \sum_{r=1}^{M} I_{ \{X_6^{(r)}=3\} }
\end{align*}

**Paso 5.** Cacule la proporción de trayctorias que terminarion en el estado $3$:

\begin{align*}
\hat{p}_{13}^{(6)}= \frac{N_3}{M}= \frac{1}{M} \sum_{r=1}^{M} I_{ \{X_6^{(r)}=3\} }
\end{align*}

esta proporción es la estimación de la probabilidad buscada.


#### Cálculo teórico de $p^{(6)}_{13}$, paso a paso

**Idea.** Por la ecuación de Chapman–Kolmogorov, la probabilidad de pasar de $i$ a $j$ en $n$ pasos es la entrada $(i,j)$ de $P^n$:
$$
\mathbb{P}(X_6=3\mid X_0=1)=p^{(6)}_{13}=(P^6)_{13}.
$$
En vez de multiplicar $P$ seis veces, usamos $P^6=P^3\,P^3$. Así basta con calcular $P^3$ y multiplicar la **fila 1** de $P^3$ por la **columna 3** de $P^3$:
$$
p^{(6)}_{13}=\sum_{k=1}^{4}p^{(3)}_{1k}\,p^{(3)}_{k3}.
$$

Recordemos que cada entrada de un producto de matrices es fila por columna: $(AB)_{ij}=\sum_{k=1}^4 A_{ik}B_{kj}$. En código, el producto de matrices es `P @ P` (o `numpy.matmul`/`numpy.dot`). **Ojo:** `numpy.multiply(P, P)` multiplica entrada por entrada ($P_{ij}^2$), que **no** es $P^2$.

```python
P = numpy.array([
    [0.70, 0.20, 0.05, 0.05],
    [0.30, 0.40, 0.20, 0.10],
    [0.05, 0.10, 0.55, 0.30],
    [0.40, 0.20, 0.10, 0.30]
])

# Producto matricial (fila × columna). numpy.multiply sería entrada por entrada y daría otra cosa.
P2 = P @ P
print("P^2 = P·P =")
print(P2)

P3 = P2 @ P
print("\nP^3 = P^2·P =")
print(P3)
```

**Paso final: $p^{(6)}_{13}=(P^3P^3)_{13}$.** Solo necesitamos la fila 1 y la columna 3 de $P^3$:

- Fila 1 de $P^3$: $\big(p^{(3)}_{11},p^{(3)}_{12},p^{(3)}_{13},p^{(3)}_{14}\big)=(0.510625,\ 0.23625,\ 0.14325,\ 0.109875)$: probabilidades de ir de 1 a cada estado $k$ en 3 meses.
- Columna 3 de $P^3$: $\big(p^{(3)}_{13},p^{(3)}_{23},p^{(3)}_{33},p^{(3)}_{43}\big)=(0.14325,\ 0.20375,\ 0.265625,\ 0.165)$: probabilidades de ir de cada estado $k$ a 3 en otros 3 meses.

Cada término $p^{(3)}_{1k}\,p^{(3)}_{k3}$ es la probabilidad de estar en el estado $k$ en el mes 3 y llegar a 3 en el mes 6. Sumando sobre $k$ se obtienen todos los caminos posibles.

```python
fila_1 = P3[0, :]   # fila 1 de P^3
col_3  = P3[:, 2]   # columna 3 de P^3

print("Fila 1 de P^3:   ", fila_1)
print("Columna 3 de P^3:", col_3)

# Fila × columna: se multiplican término a término y se suman
terminos = fila_1 * col_3
print("Términos p^(3)_1k · p^(3)_k3:", terminos)

p13_6 = numpy.sum(terminos)          # equivalente a fila_1 @ col_3
print(f"\np^(6)_13 = {p13_6:.9f}")

# Verificación con la potencia de la matriz
print(f"Verificación (P^6)_13 = {numpy.linalg.matrix_power(P, 6)[0, 2]:.9f}")
```

```python
# Definir condiciones
M = 1000
N = 6
X_0 = 1

def ejercicio5(M, N, X_0):
  N_3 = 0
  for r in range(M):
    camino = trayectory(X_0, N)
    x_6 = camino[-1]
    if x_6 == 3:
      N_3 += 1
  return N_3

N_3 = ejercicio5(M, N, X_0)

p_estimada = N_3 / M

print(f"Número de trayectorias que terminaron en estado 3: {N_3} de {M}")
print(f"Probabilidad estimada p_13^(6): {p_estimada:.4f}")
```

### Ejercicio 6a.
Determinte si la cadena es irreducible y aperiódica. Calcule una distribución estacionaria $\pi$ de la cadena, solucionando $\pi P = \pi$, ¿es única?

**Irreducibilidad y aperiodicidad.** Todas las entradas de $P$ son estrictamente positivas ($P_{ij}\ge 0.05>0$). Entonces:

- **Irreducible:** desde cualquier estado $i$ se llega a cualquier estado $j$ en un solo paso ($P_{ij}>0$), así que todos los estados se comunican.
- **Aperiódica:** $P_{ii}>0$ para todo $i$, luego $1\in\{n\ge1: P^{(n)}_{ii}>0\}$ y el periodo de cada estado es $\gcd=1$.

**Distribución estacionaria.** Se resuelve $\pi P=\pi$, $\sum_i\pi_i=1$, es decir $(P^T-I)\pi^T=0$ junto con la condición de normalización. La solución exacta es
$$
\pi=\left(\frac{270}{603},\ \frac{137}{603},\ \frac{110}{603},\ \frac{86}{603}\right)\approx(0.447761,\ 0.227197,\ 0.182421,\ 0.142620).
$$

**¿Es única?** Sí. Una cadena finita, irreducible y aperiódica tiene una única distribución estacionaria (Häggström, cap. 5), y además $\mu^{(n)}\to\pi$ para cualquier distribución inicial. Equivalentemente: como la cadena es irreducible, $\lambda=1$ es valor propio simple de $P^T$, así que el sistema anterior tiene solución única.

#### ¿La solución de $\pi P=\pi$ es única salvo constantes?

Sin la condición $\sum_i\pi_i=1$, la ecuación $\pi P=\pi$ es un sistema **homogéneo**: $\pi(P-I)=0$, o equivalentemente $(P^T-I)\,\pi^T=0$. Escrito completo:

$$
\begin{aligned}
-0.30\,\pi_1+0.30\,\pi_2+0.05\,\pi_3+0.40\,\pi_4&=0\\
 0.20\,\pi_1-0.60\,\pi_2+0.10\,\pi_3+0.20\,\pi_4&=0\\
 0.05\,\pi_1+0.20\,\pi_2-0.45\,\pi_3+0.10\,\pi_4&=0\\
 0.05\,\pi_1+0.10\,\pi_2+0.30\,\pi_3-0.70\,\pi_4&=0
\end{aligned}
$$

Si $\pi$ es solución, también lo es $c\,\pi$ para cualquier constante $c$, así que nunca hay una sola solución. La pregunta es si **todas** las soluciones son múltiplos de un mismo vector, es decir, si el conjunto solución (el núcleo de $P^T-I$) tiene **dimensión 1**.

Observación: las cuatro ecuaciones suman $0$, porque cada fila de $P$ suma $1$ y entonces cada columna de $P^T-I$ suma $0$. Por eso una ecuación sobra y el rango es a lo sumo $3$. Hay que comprobar que es exactamente $3$; así, por el teorema rango–nulidad, el núcleo tiene dimensión $4-3=1$.

Lo hacemos con eliminación de Gauss en aritmética exacta (fracciones), para que no haya errores de redondeo.

```python
from fractions import Fraction

# Matriz P^T - I con fracciones exactas
Pf = [[Fraction(round(x * 100), 100) for x in fila] for fila in P]
A = [[Pf[j][i] - (1 if i == j else 0) for j in range(4)] for i in range(4)]

def mostrar_frac(M, titulo):
    print(titulo)
    for fila in M:
        print("  [" + "  ".join(f"{str(x):>9}" for x in fila) + " ]")
    print()

mostrar_frac(A, "P^T - I =")

# Eliminación de Gauss-Jordan (forma escalonada reducida)
R = [fila[:] for fila in A]
fila_pivote = 0
columnas_pivote = []
for col in range(4):
    candidatas = [r for r in range(fila_pivote, 4) if R[r][col] != 0]
    if not candidatas:
        continue
    r = candidatas[0]
    R[fila_pivote], R[r] = R[r], R[fila_pivote]
    piv = R[fila_pivote][col]
    R[fila_pivote] = [x / piv for x in R[fila_pivote]]
    for k in range(4):
        if k != fila_pivote and R[k][col] != 0:
            factor = R[k][col]
            R[k] = [a - factor * b for a, b in zip(R[k], R[fila_pivote])]
    columnas_pivote.append(col)
    fila_pivote += 1

mostrar_frac(R, "Forma escalonada reducida de P^T - I =")

rango = len(columnas_pivote)
libres = [c for c in range(4) if c not in columnas_pivote]
print(f"Rango de P^T - I: {rango}")
print(f"Dimensión del núcleo (4 - rango): {4 - rango}")
print(f"Variable libre: pi_{libres[0] + 1}")
```

La forma escalonada reducida tiene la forma

$$
\begin{pmatrix}
1 & 0 & 0 & -a\\
0 & 1 & 0 & -b\\
0 & 0 & 1 & -c\\
0 & 0 & 0 & 0
\end{pmatrix},
$$

con una fila de ceros (la ecuación redundante) y $\pi_4$ como única variable libre. Despejando $\pi_1=a\,\pi_4$, $\pi_2=b\,\pi_4$, $\pi_3=c\,\pi_4$, todas las soluciones son

$$
\pi = \pi_4\,(a,\ b,\ c,\ 1),\qquad \pi_4\in\mathbb{R}.
$$

La siguiente celda calcula $a,b,c$ y escribe el vector con enteros.

```python
from math import lcm

# Solución general: pi_j = -R[j][3] * pi_4 para j = 1, 2, 3 y pi_4 libre
v = [-R[j][3] for j in range(3)] + [Fraction(1)]
print("Solución general: pi = t ·", [str(x) for x in v], ", con t real")

# Reescalar para tener enteros
m = lcm(*[x.denominator for x in v])
v_ent = [int(x * m) for x in v]
print("Equivalentemente: pi = s ·", v_ent, ", con s real")

# Comprobación: v·P = v, y también (c·v)·P = c·v para cualquier c
vP = [sum(v_ent[i] * Pf[i][j] for i in range(4)) for j in range(4)]
print("\nv·P =", [str(x) for x in vP], "  (igual a v)")
for c in [2, -3, Fraction(1, 7)]:
    cv = [c * x for x in v_ent]
    cvP = [sum(cv[i] * Pf[i][j] for i in range(4)) for j in range(4)]
    print(f"c = {str(c):>4}:  (c·v)·P == c·v ? {cvP == cv}")

# La normalización fija la constante
s = Fraction(1, sum(v_ent))
pi_exacta = [s * x for x in v_ent]
print(f"\nSuma de v = {sum(v_ent)}  ->  única constante que normaliza: s = {s}")
print("pi =", [str(x) for x in pi_exacta])
print("pi ≈", [round(float(x), 6) for x in pi_exacta])
```

**Otra forma de verlo: valores propios.** $\pi P=\pi$ dice que $\pi^T$ es vector propio de $P^T$ con valor propio $1$. Si $\lambda=1$ es un valor propio **simple**, su espacio propio tiene dimensión $1$ y $\pi$ es única salvo constantes.

```python
valores, vectores = numpy.linalg.eig(P.T)
print("Valores propios de P^T:", numpy.round(valores, 6))
print("Módulos:               ", numpy.round(numpy.abs(valores), 6))
print("Multiplicidad de lambda = 1:", int(numpy.sum(numpy.isclose(valores, 1))))
print("Rango de P^T - I (numpy):", numpy.linalg.matrix_rank(P.T - numpy.eye(4)))

# Vector propio asociado a lambda = 1, normalizado para que sume 1
idx = numpy.argmin(numpy.abs(valores - 1))
w = numpy.real(vectores[:, idx])
print("\nVector propio de lambda = 1 (salida de numpy):", numpy.round(w, 6))
print("Dividido por su suma:                       ", numpy.round(w / w.sum(), 6))
```

**Conclusión.** El rango de $P^T-I$ es $3$, así que el núcleo tiene dimensión $1$: toda solución de $\pi P=\pi$ es de la forma

$$
\pi = s\,(270,\ 137,\ 110,\ 86),\qquad s\in\mathbb{R}.
$$

Es decir, **la solución es única salvo una constante multiplicativa**. Lo mismo dicen los valores propios: $\lambda=1$ es simple y los demás tienen módulo menor que $1$. Al imponer $\sum_i\pi_i=1$, la constante queda fijada en $s=1/603$, y la distribución estacionaria es **única**:

$$
\pi=\left(\tfrac{270}{603},\ \tfrac{137}{603},\ \tfrac{110}{603},\ \tfrac{86}{603}\right)\approx(0.447761,\ 0.227197,\ 0.182421,\ 0.142620).
$$

Esto es consistente con la teoría: una cadena finita e irreducible tiene distribución estacionaria única.

Calcule las frecuencias reltavias de visita a cada estado presente sus resultados en la tabla.

\begin{array}{c c c c c}
X_0 & \text{Estado 1} & \text{Estado 2} & \text{Estado 3} & \text{Estado 4}\\

\end{array}

Compare los resultados y responda: ¿Depende la distribución de las frecuencias relativas a largo plazo del estado inicial? Justifique su respuesta.

```python
N=10000
print(f"{'X_0':>5} | {'Estado 1':>10} | {'Estado 2':>10} | {'Estado 3':>10} | {'Estado 4':>10}")
print("-" * 57)

for x_0 in [1,2,3,4]:
  camino = trayectory(x_0, N)
  frec = [camino.count(j) / (N+1) for j in [1,2,3,4]]
  print(f"{x_0:>5} | {frec[0]:>10.6f} | {frec[1]:>10.6f} | {frec[2]:>10.6f} | {frec[3]:>10.6f}")

print("-" * 57)
print(f"{'π':>5} | {pi[0]:>10.6f} | {pi[1]:>10.6f} | {pi[2]:>10.6f} | {pi[3]:>10.6f}")
```

**Respuesta.** No, la distribución de las frecuencias relativas a largo plazo **no depende del estado inicial**. Las cuatro filas de la tabla son muy parecidas entre sí y todas están cerca de $\pi$; las diferencias que se ven son fluctuaciones aleatorias que se reducen al aumentar la longitud de la trayectoria.

La justificación es teórica: la cadena es finita, irreducible y aperiódica, así que tiene una única distribución estacionaria $\pi$ y, por el teorema ergódico, para **cualquier** estado inicial $X_0=i$,
$$
Y_j(n)=\frac{1}{n+1}\sum_{m=0}^n I_{\{X_m=j\}}\xrightarrow[n\to\infty]{}\pi_j\quad\text{c.s.}
$$
El estado inicial solo afecta los primeros meses, y su efecto se diluye al promediar sobre $10000$ pasos.
